```python
df=pd.read_csv('Multiple_logistic_Regression.csv',sep=';')
print(df.head())
```

```python
x=df[['Study_Hours','Attendance_Percent','Assignments_Completed','Previous_Score','Sleep_Hours']]

y=df['Pass']
print(x.head())
print(y.head())
```

```python
print(df.describe())
```

```python
import sklearn.metrics as metrics
from sklearn.linear_model import LogisticRegression
import statsmodels.api as sm
from sklearn.metrics import confusion_matrix
```

```python
logistoc_model=sm.Logit(y,x).fit()
x=sm.add_constant(x)
```

```python
print(logistoc_model.summary())
```

```python

y_pred= logistoc_model.predict(x)
print(pd.DataFrame({
    "y_actual": y,
    "t_pred":y_pred
}))
```

```python
y_pred_binary = (y_pred > 0.5).astype(int)
cofussion=confusion_matrix(y, y_pred_binary)
print(cofussion)
```

```python
print(metrics.accuracy_score(y,y_pred_binary))
print(metrics.precision_score(y,y_pred_binary))
print(metrics.recall_score(y,y_pred_binary))
print(metrics.f1_score(y,y_pred_binary))
```
